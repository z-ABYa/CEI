# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (COMPLETED)

import re

def agent(query: str):
    """
    Smart single-agent pipeline.

    Routing Logic:
    1. Calculation requests -> Calculator Tool
    2. Keyword extraction requests -> Keyword Tool
    3. Other queries -> General response

    Returns:
        dict: Structured JSON-like response.
    """

    # Basic input validation
    if not isinstance(query, str) or not query.strip():
        return {
            "type": "error",
            "result": "Please provide a valid query."
        }

    query_lower = query.lower().strip()

    try:
        # -------------------------------
        # Route 1: Calculator
        # -------------------------------
        if "calculate" in query_lower:
            # Extract mathematical expression after the word "calculate"
            expression = re.sub(r"calculate\s*", "", query, flags=re.IGNORECASE).strip()

            return {
                "type": "calculation",
                "result": calculator(expression)
            }

        # -------------------------------
        # Route 2: Keyword Extraction
        # -------------------------------
        elif "keyword" in query_lower:
            # Remove helper words before extracting keywords
            text = re.sub(
                r"extract\s+keywords\s+from\s*",
                "",
                query,
                flags=re.IGNORECASE,
            ).strip()

            return {
                "type": "keywords",
                "result": extract_keywords(text)
            }

        # -------------------------------
        # Route 3: General Query
        # -------------------------------
        else:
            return {
                "type": "general",
                "result": f"I understood your query: '{query}'. No external tool was required."
            }

    except Exception as e:
        # Generic error handler
        return {
            "type": "error",
            "result": str(e)
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```
